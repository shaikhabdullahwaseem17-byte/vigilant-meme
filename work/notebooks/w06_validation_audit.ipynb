```python
# 3) Leakage audit: find suspicious columns and test effect
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split

ROOT = Path('.')
prepared = ROOT / 'data' / 'processed' / 'refresh_feature_vector.csv'
if not prepared.exists():
    raise SystemExit("Prepared features missing. Run scripts/01_prepare_features.py first.")
df = pd.read_csv(prepared)
target = 'is_declining_label'
df[target] = df.get(target, df.get('trend_direction','').astype(str).str.lower().eq('down').astype(int))

# 1) obvious label-derived columns
suspects = [c for c in df.columns if 'trend' in c.lower() or c.lower().endswith('_label') or 'declin' in c.lower()]
print("Suspect label-derived columns:", suspects)

# 2) high correlation with label (quick scan)
corrs = {}
for c in df.select_dtypes(include=[np.number]).columns:
    if c == target:
        continue
    try:
        corr = abs(df[[c, target]].dropna()[c].corr(df[target]))
        corrs[c] = corr
    except Exception:
        corrs[c] = 0.0
# show top correlated numeric features
top_corr = sorted(corrs.items(), key=lambda x: x[1], reverse=True)[:10]
print("\nTop correlations with target (abs):")
for name, val in top_corr:
    print(f"{name}: {val:.3f}")

# 3) Train-without / train-with test for immediate leakage check
numeric_feats = [c for c in df.select_dtypes(include=[np.number]).columns if c != target]
# ensure we have something
if not numeric_feats:
    print("No numeric features to test for leakage.")
else:
    # sample small holdout
    train_idx, test_idx = train_test_split(np.arange(len(df)), test_size=0.25, random_state=42, stratify=df[target])
    X_train = df.iloc[train_idx][numeric_feats].fillna(0).values
    y_train = df.iloc[train_idx][target].values
    X_test = df.iloc[test_idx][numeric_feats].fillna(0).values
    y_test = df.iloc[test_idx][target].values

    clf = LogisticRegression(max_iter=2000, random_state=42).fit(X_train, y_train)
    probs = clf.predict_proba(X_test)[:,1]
    auc_full = roc_auc_score(y_test, probs) if len(np.unique(y_test))>1 else np.nan
    print(f"\nAUC using all numeric features: {auc_full:.3f}")

    # if suspects present, drop them and re-evaluate
    suspects_in_feats = [s for s in suspects if s in numeric_feats]
    if suspects_in_feats:
        X_train2 = df.iloc[train_idx][[c for c in numeric_feats if c not in suspects_in_feats]].fillna(0).values
        X_test2 = df.iloc[test_idx][[c for c in numeric_feats if c not in suspects_in_feats]].fillna(0).values
        clf2 = LogisticRegression(max_iter=2000, random_state=42).fit(X_train2, y_train)
        probs2 = clf2.predict_proba(X_test2)[:,1]
        auc_no_suspects = roc_auc_score(y_test, probs2) if len(np.unique(y_test))>1 else np.nan
        print(f"AUC after removing suspect columns {suspects_in_feats}: {auc_no_suspects:.3f}")
        print("If AUC collapses dramatically after removal, you likely had label leakage.")
    else:
        print("No explicit suspect columns (trend_* or *_label) found among numeric features.")

# 4) window overlap quick check: look for *_prev* vs *_last* columns
overlap_cols = [c for c in df.columns if any(s in c for s in ['prev30', 'prev_30', 'last_30', 'last30', '_90d'])]
print("\nColumns that may indicate overlapping windows (inspect manually):", overlap_cols)

# 5) Decision-derived product flags check (columns that look like actions/scores)
product_flags = [c for c in df.columns if any(s in c for s in ['suggested_action','baseline_refresh_score','reason_codes','suggested_action_baseline'])]
print("Product-flag-like columns present (should NOT be used as features):", product_flags)

# Guidance summary to paste into the markdown above:
print("\nGuidance to paste into markdown:")
print("- If any 'trend' columns or explicit label columns are present in your features, remove them and re-run evaluations.")
print("- If AUC drops a lot when removing suspects, you had leakage — report the leak and re-evaluate honestly.")
print("- Avoid using product flags or existing decision scores as inputs; treat them as baselines instead.")
```

```python
# Fixed rerun-evaluations cell: drops suspects, runs splits, shows FP/FN safely
import numpy as np, pandas as pd, os, sys
from pathlib import Path
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import train_test_split, GroupShuffleSplit
from sklearn.preprocessing import StandardScaler
from IPython.display import display

# helper from repo
from scripts import ml_utils
from scripts.ml_utils import percentile_rank, normalize, precision_at_k

ROOT = Path.cwd()
prepared = ROOT / 'data' / 'processed' / 'refresh_feature_vector.csv'
df = pd.read_csv(prepared)
df['is_declining_label'] = df.get('is_declining_label', df.get('trend_direction','').astype(str).str.lower().eq('down').astype(int))

suspects = [c for c in df.columns if c.lower().startswith('trend') or c.lower().endswith('_label') or 'declin' in c.lower()]
exclude_ids = ['content_id','client_id','is_declining_label']
suspects = list(set(suspects + exclude_ids))
print("Dropping suspect/ID/label columns:", suspects)

numeric_cols = [c for c in df.select_dtypes(include=[np.number]).columns if c not in suspects]
print("Candidate numeric features (count):", len(numeric_cols))
print("Using numeric features (first 20):", numeric_cols[:20])

def eval_split(train_idx, test_idx, name):
    train = df.iloc[train_idx]
    test = df.iloc[test_idx].copy()
    X_train = train[numeric_cols].fillna(0).values
    X_test = test[numeric_cols].fillna(0).values
    y_train = train['is_declining_label'].values
    y_test = test['is_declining_label'].values

    scaler = StandardScaler().fit(X_train)
    X_train_s = scaler.transform(X_train)
    X_test_s = scaler.transform(X_test)

    clf_lr = LogisticRegression(max_iter=5000, random_state=42, solver='lbfgs').fit(X_train_s, y_train)
    clf_rf = RandomForestClassifier(n_estimators=200, random_state=42).fit(X_train, y_train)
    probs_lr = clf_lr.predict_proba(X_test_s)[:,1]
    probs_rf = clf_rf.predict_proba(X_test)[:,1]

    # baseline on test
    test['visibility_score'] = percentile_rank(np.log1p(test['impressions_90d']))
    test['freshness_risk_score'] = percentile_rank(test.get('days_since_last_update', test['content_age_days']))
    test['position_opportunity_score'] = ((1 - normalize(test['avg_position'].clip(lower=1, upper=50))) * test['visibility_score'] * (test.get('avg_position',0) > 0).astype(int))
    test['depth_gap_score'] = (1 - percentile_rank(test['word_count'])) * test['visibility_score']
    test['baseline_refresh_score'] = (0.40*test['visibility_score'] + 0.30*test['freshness_risk_score'] + 0.25*test['position_opportunity_score'] + 0.05*test['depth_gap_score']).clip(0,1)

    rows = {}
    for k in (20,50):
        rows[f'baseline_prec@{k}'] = precision_at_k(test['is_declining_label'].values, test['baseline_refresh_score'].values, k)
        rows[f'lr_prec@{k}'] = precision_at_k(test['is_declining_label'].values, probs_lr, k)
        rows[f'rf_prec@{k}'] = precision_at_k(test['is_declining_label'].values, probs_rf, k)
    rows['auc_lr'] = roc_auc_score(y_test, probs_lr) if len(np.unique(y_test))>1 else float('nan')
    rows['auc_rf'] = roc_auc_score(y_test, probs_rf) if len(np.unique(y_test))>1 else float('nan')
    rows['base_rate'] = float(np.mean(y_test))
    print(f"\n== Results for {name} (n_test={len(test)}) ==")
    for k,v in rows.items():
        print(f"{k}: {v}")

    # Show FP/FN safely
    test_with_preds = test.reset_index(drop=True)
    test_with_preds['prob_rf'] = probs_rf
    test_with_preds['prob_lr'] = probs_lr

    fp = test_with_preds[(test_with_preds['prob_rf'] >= 0.5) & (test_with_preds['is_declining_label']==0)].sort_values('prob_rf', ascending=False).head(5)
    fn = test_with_preds[(test_with_preds['prob_rf'] < 0.5) & (test_with_preds['is_declining_label']==1)].sort_values('prob_rf').head(5)

    cols_show = ['content_id','prob_rf','prob_lr','baseline_refresh_score','impressions_90d','engagement_rate','is_declining_label']
    cols_show = [c for c in cols_show if c in test_with_preds.columns]  # filter out missing
    if not cols_show:
        print("No columns available to display FP/FN.")
    else:
        print("\nTop false positives (sample):")
        display(fp[cols_show])
        print("\nTop false negatives (sample):")
        display(fn[cols_show])

    return rows, clf_lr, clf_rf

# Run stratified random split
indices = np.arange(len(df))
train_val_idx, test_idx = train_test_split(indices, test_size=0.2, random_state=42, stratify=df['is_declining_label'])
train_idx, _ = train_test_split(train_val_idx, test_size=0.25, random_state=42, stratify=df.iloc[train_val_idx]['is_declining_label'])
res_random, lr1, rf1 = eval_split(train_idx, test_idx, "stratified random")

# Grouped-by-client
if 'client_id' in df.columns:
    gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
    train_idx_g, test_idx_g = next(gss.split(df, groups=df['client_id']))
    res_grouped, lr2, rf2 = eval_split(train_idx_g, test_idx_g, "grouped-by-client")
else:
    print("No client_id; skipped grouped split")
    res_grouped = None

# Time-aware
if 'content_age_days' in df.columns:
    cutoff = df['content_age_days'].quantile(0.3)
    train_idx_time = df[df['content_age_days'] >= cutoff].index.values
    test_idx_time = df[df['content_age_days'] < cutoff].index.values
    if len(test_idx_time) > 50 and len(train_idx_time) > 50:
        res_time, lr3, rf3 = eval_split(train_idx_time, test_idx_time, "time-aware")
    else:
        print("Not enough rows for robust time split; skipped")
        res_time = None
else:
    print("No content_age_days; skipped time-aware")
    res_time = None

print("\nDone. Summary:\nrandom:", res_random, "\ngrouped:", res_grouped, "\ntime:", res_time)
```
