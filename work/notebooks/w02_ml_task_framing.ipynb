Now, let's load the dataset named `abdullehh/internship-warehouse-bucket` and inspect its structure. Hugging Face datasets are typically loaded as `DatasetDict` or `Dataset` objects.

```python
from datasets import load_dataset
import pandas as pd

dataset_name = 'FlyRank/internship-warehouse'
config_name = 'fact_content_daily_performance' # Choosing a relevant config based on the error message

try:
    # Load the dataset with the specified config
    hf_dataset = load_dataset(dataset_name, config_name)

    print(f"Successfully loaded dataset: {dataset_name} with config {config_name}")
    print("Dataset structure:")
    print(hf_dataset)

    # If the dataset is a DatasetDict (common for train/test splits), we'll try to get the 'train' split
    if isinstance(hf_dataset, dict) and 'train' in hf_dataset:
        # Convert only the first 5 rows to a pandas DataFrame for efficient display
        df_from_hf = hf_dataset['train'].select(range(5)).to_pandas()
        print("\nFirst 5 rows of the 'train' split converted to Pandas DataFrame (for demonstration):")
        display(df_from_hf.head())
        print(f"\nDataFrame shape (first 5 rows): {df_from_hf.shape}")

    elif isinstance(hf_dataset, pd.DataFrame):
        # This path is less common for direct load_dataset output but included for robustness
        print("\nFirst 5 rows of the dataset (already a Pandas DataFrame):")
        display(hf_dataset.head())
        print(f"\nDataFrame shape: {hf_dataset.shape}")

    else:
        # If it's a single Dataset object, convert only the first 5 rows
        df_from_hf = hf_dataset.select(range(5)).to_pandas()
        print("\nFirst 5 rows of the dataset converted to Pandas DataFrame (for demonstration):")
        display(df_from_hf.head())
        print(f"\nDataFrame shape (first 5 rows): {df_from_hf.shape}")

except Exception as e:
    print(f"Error loading or processing dataset '{dataset_name}' with config '{config_name}': {e}")
    print("Please ensure the dataset name and config are correct, and you have internet access.")
    print("If it's a private dataset, you might need to login to Hugging Face Hub: `from huggingface_hub import login; login()`")
```
